# ⚡ FRUMI AI VIDEO — AUTONOMOUS GPU WORKER (COLAB T4)
### HỆ THỐNG MÁY TRẠM GPU TỰ ĐỘNG KẾT NỐI VÀ SINH VIDEO 30FPS

> **👉 HƯỚNG DẪN:**
> 1. Kiểm tra **Runtime > Change runtime type > T4 GPU**.
> 2. Nhấn menu **Runtime > Run all** (hoặc **Ctrl+F9** / **Cmd+F9**).
> 3. Giữ tab này mở. Hệ thống sẽ tự động hoàn toàn!


```python
# @title 🔍 [Bước 1/2] Kiểm tra phần cứng GPU NVIDIA Tesla T4
import subprocess
res = subprocess.run(["nvidia-smi", "--query-gpu=name,memory.total", "--format=csv,noheader"], capture_output=True, text=True)
if res.returncode != 0:
    print("❌ LỖI: Chưa kích hoạt GPU! Hãy vào Runtime > Change runtime type > Chọn T4 GPU rồi chạy lại!")
    raise SystemExit(1)
print(f"✅ Đã kích hoạt GPU: {res.stdout.strip()}")
```
